```python
import sqlite3
```

```python
#Connect to the sqlite Database
#How to create a database
connection = sqlite3.connect('test.db')
connection
```

```python
#cursor Object
cursor = connection.cursor()
```

```python
#Creating a table in db

cursor.execute('''
    Create Table if Not Exists Employess(
        Id Integer Primary key,
        name text Not Null,
        age integer,
        Profession text
        )

''')

#Commiting the changes 
connection.commit()
```

```python
#Insert the Data in SQL lite Table
cursor.execute('''
    Insert into Employess(name,age,Profession)
               values('Arjun',34,'Pornstar')

''')

cursor.execute('''
    Insert into Employess(name,age,Profession)
               values('Vishnu',34,'Data Engineer')

''')

cursor.execute('''
    Insert into Employess(name,age,Profession)
               values('Rohan',22,'Architect')

''')

connection.commit()
```

```python
#QUery the data from the table
cursor.execute('SELECT * FROM Employess')

rows = cursor.fetchall()

#print the all data
for row in rows :
    print(row)
```

```python
#UPdat the data in the Table

cursor.execute('''
UPDATE Employess
SET name='Adit' , Profession = 'Analyst'
WHERE Id = 2

''')

connection.commit()
```

```python
#Delete the Data from the Table 

cursor.execute('''
DELETE from Employess 
               where id=2
''')
connection.commit()

```
