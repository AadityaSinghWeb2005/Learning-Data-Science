```python
#Color , linestyle,marker,labels,legend,grid

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
visitors = [120, 150, 90, 300, 280, 400, 350]

plt.plot(
    days,
    visitors,
    color="green",
    marker="o",
    linestyle="--",
    label="Website Visitors"
)

plt.xlabel("Days")
plt.ylabel("Visitors Count")
plt.title("Weekly Website Traffic Analysis")

plt.legend()
plt.grid(True)

plt.show()
```

```python
#Multiple Lines in One graph

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
visitors = [120, 150, 90, 300, 280, 400, 350]
orders = [30, 45, 20, 80, 75, 120, 100]

plt.plot(
    days,
    visitors,
    color="blue",
    marker="o",
    linestyle="-",
    label="Visitors"
)

plt.plot(
    days,
    orders,
    color="red",
    marker="^",
    linestyle="--",
    label="Orders"
)

plt.xlabel("Days")
plt.ylabel("Count")
plt.title("Visitors vs Orders (Weekly Analysis)")

plt.legend()
plt.grid(True)

plt.show()
```

```python
#Bar Chart

#Vertical Bar Chart
import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet", "Watch"]
revenue = [50000, 70000, 30000, 20000]

plt.bar(products, revenue, color="orange")

plt.xlabel("Products")
plt.ylabel("Revenue")
plt.title("Product-wise Revenue Analysis")

plt.show()
```

```python
#Scatter Plot

import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 6, 7, 8]
marks = [40, 45, 50, 55, 65, 70, 78, 85]

plt.scatter(
    study_hours,
    marks,
    color='blue'
)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")

plt.show()
```

```python
#Multiple Charts in One graph

#Same data - Different perspectives side by side
#Ek single Canvas ke andar multiple CHart fitin karna

#CODE:
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 6, 7, 8]
marks = [40, 45, 50, 55, 65, 70, 78, 85]

fig,axes = plt.subplots(1,2,figsize=(12,4))

#plot-1 LIne plot
axes[0].plot(study_hours,marks,color='blue',marker='o')
axes[0].set_title('study Hours vs marks')
axes[0].set_xlabel('Study Hours')
axes[0].set_ylabel('Marks')
axes[0].grid(True)

#plot 2 : Histogram
axes[1].hist(marks,color='red',edgecolor='black')
axes[1].set_title('Marks Distribution')
axes[1].set_xlabel('Marks range')
axes[1].set_ylabel('No. of Student')
axes[1].grid(True)

plt.tight_layout()#Overlapping Text fix karta hai and spacing clean karta hai
plt.show()
```

```python
#Exercise(Multiple chart in one graph)

import matplotlib.pyplot as plt

cities =['Delhi','Jaipur','Pune','Mumbai','Kota']
populationData =[20,22,13,27,29]

fig,axes = plt.subplots(1,2,figsize=(12,4))

#Plot -1 : Bar Chart
axes[0].bar(cities,populationData,color='Green',label='Population')
axes[0].set_title('Population Data of Cities')
axes[0].set_xlabel('Cities')
axes[0].set_ylabel('Population')
axes[0].legend()
axes[0].grid(True)

#Plot-2 : Histogram
axes[1].hist(populationData,color='blue',label='Population')
axes[1].set_title('Population Distribution')
axes[1].set_xlabel('Population Data')
axes[1].set_ylabel('no. of cities')
axes[1].legend()
axes[1].grid(True)

plt.tight_layout()
plt.show()

```

```python
#Firgure size , DPI & Styling(Professional Look )
'''
Notes :
plt.figures(figsize=(8,4))
8-> width (inches)
4 -> height(inches)

DPI = Dot per inch
DPI	Use
80	Rough / testing
100–120	Notebook / GitHub ✅
150–200	Presentation / Report

industry default dpi=120
“dpi improves resolution for reports and GitHub.”
'''

#CODE :

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
visitors = [120, 150, 90, 300, 280, 400, 350]

plt.figure(figsize=(8,4),dpi=120)

plt.plot(
    days,
    visitors,
    color='red',
    marker='o',
    label='visitors'
)
plt.xlabel('Days')
plt.ylabel('No. of visitors')
plt.title('No. of visitors on every weekday ')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
