```python

#Exercise

import seaborn as sns
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [200, 250, 180, 300, 350, 400]

sns.set_style('darkgrid')
plt.figure(figsize=(8,4),dpi=120)

sns.lineplot(x=months,y=sales,marker ="o",label='Sales')
plt.title('Monthly Sales trend')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.legend()
plt.show()
```

```python
#Distribution Plot(hist,kDE)
'''
Notes :
Distribution plot batata hai :
  1. Data kitna spread horha hai
  2. values kis range mein zada hai
  3. Data Normal/skewed hai ya nahi

'''
#histplot
#CODE:
import seaborn as sns
import matplotlib.pyplot as plt

marks = [45, 50, 55, 60, 62, 65, 68, 70, 72, 75, 78, 80, 85, 90, 95]

sns.set_style('whitegrid')
plt.figure(figsize=(8,4),dpi=120)

sns.histplot(marks,bins=5)
plt.title('Marks Distribution')
plt.xlabel('Marks')
plt.ylabel('Number of Students')

plt.show()
```

```python
#KDE Plot(Used for distribution)

'''
Histogram ka smooth curve version
Peaks clearly dikhta hai
'''
import seaborn as sns
import matplotlib.pyplot as plt

marks = [45, 50, 55, 60, 62, 65, 68, 70, 72, 75, 78, 80, 85, 90, 95]

sns.set_style('whitegrid')
plt.figure(figsize=(8,4),dpi=120)

sns.histplot(marks,bins=5,kde=True)
plt.title("Marks Distribution with KDE")
plt.xlabel("Marks")
plt.ylabel("Count")

plt.show()
```

```python
#Categorical Plots
'''
Categorical data yeh plots use hote hai :
  1. Data categories me ho
  2. Hume compare/count/spread dekhna ho

  Examples :
    Gender wise count
    Product wise sales
    Category wise sales
    Outliers detection
'''

#1. Countplot

#Use : kis cateogry mein kitne Records hai Eg. Kitene male/female employess
import seaborn as sns
import matplotlib.pyplot as plt

gender = ['Male', 'Female', 'Male', 'Female', 'Male', 'Male', 'Female']

sns.set_style('whitegrid')
plt.figure(figsize=(6, 4), dpi=120)

sns.countplot(x=gender)

plt.title('Gender Count')
plt.xlabel('Gender')
plt.ylabel('Count')

plt.show()
```

```python
#2. Barplot
'''
countplot --> sirf count
barplot --> mean/aggregation

Example : department wise average salary
'''
import seaborn as sns
import matplotlib.pyplot as plt

department = ['HR', 'IT', 'IT', 'HR', 'Sales', 'Sales']
salary = [30, 50, 60, 35, 45, 55]

plt.figure(figsize=(6, 4), dpi=120)

sns.barplot(x=department, y=salary)

plt.title('Average Salary by Department')
plt.xlabel('Department')
plt.ylabel('Salary')

plt.show()
```

```python
#3. boxPlot

'''
its shows Distribution + OUtliers

BoxPlot dikhata hai :
  1. Median
  2. Quartiles
  3. spread
  4. Outliers
'''
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4), dpi=120)

sns.boxplot(x=department, y=salary)

plt.title('Salary Distribution by Department')
plt.xlabel('Department')
plt.ylabel('Salary')

plt.show()

#

```

```python
# 4. violinplot - Boxplot + KDE

'''
Boxplot + Density
its shows distribution shape


🧠 Boxplot vs Violinplot (Interview 🔥)

Boxplot	     Violinplot
Quartiles	   Density shape
Simple	     More detail
Outliers     clear	Distribution clear
'''

plt.figure(figsize=(6, 4), dpi=120)

sns.violinplot(x=department, y=salary)

plt.title('Salary Distribution (Violin Plot)')
plt.xlabel('Department')
plt.ylabel('Salary')

plt.show()
```

```python
#Exercise (countplot , barplot , boxplot)

'''
Important Points :
📌 Countplot:

Category B ke records sabse zyada

📌 Barplot:

Category C ka average value highest

📌 Boxplot:

Category B ka spread wide

Category C ka median highest

No major outliers

📌 Interview-ready line:

“I used countplot, barplot, and boxplot together to analyze frequency, central tendency, and distribution.”
'''

import seaborn as sns
import matplotlib.pyplot as plt

category = ['A', 'A', 'B', 'B', 'B', 'C', 'C']
values = [10, 15, 20, 22, 25, 30, 35]

sns.set_style('whitegrid')
plt.figure(figsize=(8,4),dpi=120)

# sns.countplot(x=category,label='Category')
# sns.barplot(x=category,y=values,label='Cateogry Mean')
sns.boxplot(x=category,y=values,label='value')


plt.title('Value per category')
plt.xlabel('Category')
plt.ylabel('values')
plt.show()
```

```python
#Heatmap and corelation matrix

#Heatmap
'''
Heatmap is used for representing data with the help of colors
zyada value --> dark color
kam value --> light color

in simple words : Number ko color mein badal dena taki patter clear dikhe

Heatmap ka common use
  1. Heatmap ka use Correlation mei kiya jata hai
  Correlation batata hai:

      Do variables ka relationship

      Value range: -1 to +1

      Value	Meaning
      +1	Perfect positive relation
      0	No relation
      -1	Perfect negative relation


'''
#CODE  :
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd


data = {
    'Age': [22, 25, 30, 35, 40],
    'Salary': [25, 30, 45, 60, 80],
    'Experience': [1, 2, 5, 7, 10]
}

df = pd.DataFrame(data)

#Correlation Matrix
corr = df.corr()
# print(corr)

#Heatmap

sns.set_style('whitegrid')
plt.figure(figsize=(8,4),dpi=120)

sns.heatmap(
    corr,
    annot=True,#values show karta hai
    cmap ='coolwarm',#color theme
    linewidth =0.5,  # cell seperation
    fmt = '.2f' #decimal format
)

plt.title('corelation heatmap')
plt.show()
'''
🧠 Heatmap se kya bolna chahiye? (VERY IMPORTANT)

Tum confidently bol sakte ho:

Age & Salary → strong positive correlation

Experience & Salary → very strong positive

Sab values +ve → direct relationship

📌 Interview-ready line:

“Heatmap helps quickly identify strong and weak correlations using color intensity.”
'''


```

```python
#Exercise (Heatmap)

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data = {
    'Maths': [60, 70, 80, 90, 85],
    'Science': [65, 75, 82, 88, 90],
    'English': [55, 60, 65, 70, 75]
}

df = pd.DataFrame(data)
corr = df.corr()
# print(corr)#correlation matrix

#Heatmap

sns.set_style('whitegrid')
plt.figure(figsize=(8,4),dpi=120)

sns.heatmap(
    corr,
    annot=True,#values show karta hai
    cmap ='coolwarm',#color theme
    linewidths =0.5,  # cell seperation
    fmt = '.2f' #decimal format

)

plt.title('Correlation b/w subjects')
plt.show()
'''
🧠 EDA INSIGHTS (MOST IMPORTANT 🔥)

Heatmap dekh kar tum confidently bol sakte ho:

Maths & Science → strong positive correlation

Science & English → moderate positive correlation

All subjects positively correlated

📌 Interview-ready line:

“Correlation heatmap helps quickly identify relationships between numerical features.”
'''
```

```python
#Relationship Plots (scatterplot,lineplot,regplot)

#1.scatterplot - Relationship + outliers

'''
Kab use karte hai ?
  1. Do numerical variables
  2. Outliers detect karne
  3. Correlation intuition


🧠 Insight

    Study hours ↑ → marks ↑

    Positive relationship

    No major outliers

📌 Interview line:

“Scatterplot helps identify relationships and outliers.”

'''

import seaborn as sns
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 6, 7, 8]
marks = [40, 45, 50, 55, 65, 70, 78, 85]

sns.set_style('whitegrid')
plt.figure(figsize=(6,4), dpi=120)

sns.scatterplot(x=study_hours, y=marks)

plt.title('Study Hours vs Marks')
plt.xlabel('Study Hours')
plt.ylabel('Marks')

plt.show()
```

```python
#2. lineplot - Trend over time / Order

'''
1. Automatic confidence interval
2. Clean styling
3. Direct dataframe support
'''

#Example MOnthly sales code

import seaborn as sns
import matplotlib.pyplot as plt

months = ['Jan','Feb','Mar','Apr','May','Jun']
sales = [200, 250, 180, 300, 350, 400]

plt.figure(figsize=(6,4), dpi=120)

sns.lineplot(x=months, y=sales, marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')

plt.show()
```

```python
#3. Regline(Relationship + Trend line)

'''
1. Scatter + Regression Line
2. Trend clearly Dikhta hai
3. Correlation intuition strong

simple words : scatter plot mein trend line
'''

#Example : Experience vs salary

experience = [1, 2, 3, 5, 7, 10]
salary = [20, 25, 30, 45, 60, 80]

plt.figure(figsize=(6,4), dpi=120)

sns.regplot(x=experience, y=salary)

plt.title('Experience vs Salary')
plt.xlabel('Experience (Years)')
plt.ylabel('Salary')

plt.show()
```

```python
#Pairplot

'''
Har numerical features ka relationship dikhata hai
Diagnol mein distribution
off diagnols mien scatter plot

in My words:
Pair plot har feature ek dusre se Relation dikhata hai Plots bana bana kar


🧠 Pairplot se kya samajh aata hai?

        Tum graph dekh kar bol sakte ho:

        Maths ↔ Science → strong positive relation

        English ↔ Maths → moderate relation

        Diagonal plots → distribution of each subject

        No extreme outliers

📌 Interview-ready line:

“Pairplot gives a quick overview of relationships and distributions in the dataset.”

'''

#CODE :
import pandas as pd

data = {
    'Maths': [60, 70, 80, 90, 85],
    'Science': [65, 75, 82, 88, 90],
    'English': [55, 60, 65, 70, 75]
}

df = pd.DataFrame(data)
df['Result'] = ['Pass', 'Pass', 'Pass', 'Fail', 'Pass']

sns.set_style('whitegrid')
sns.pairplot(
    df,
    corner = True,
    diag_kind='kde',
    hue='Result'#iss category ke hissab se color


    )
plt.show()
```

```python
#Exercise(pairplot)

import pandas as pd

data = {
    'Age': [22, 25, 30, 35, 40],
    'Salary': [25, 30, 45, 60, 80],
    'Experience': [1, 2, 5, 7, 10]
}

df = pd.DataFrame(data)


sns.set_style('whitegrid')
plt.figure(figsize=(8,4),dpi=120)

sns.pairplot(
    df,
    corner = True,
    diag_kind='kde'
    )
'''
🧠 EDA INSIGHTS (MOST IMPORTANT 🔥)

Pairplot dekh kar tum confidently bol sakte ho:

Experience ↔ Salary → strong positive relationship

Age ↔ Salary → positive trend

Diagonal KDE → data smooth & well-distributed

No extreme outliers

📌 Interview-ready line:

“I use pairplot at the start of EDA to quickly understand feature relationships and distributions.”
'''
plt.show()
```

```python
#Catplot
'''
count ,bar ,box violin - sab ek hi function mein
Facet(row/col) support

Use :
jab multiple categorial comparisons chaiye ho tab
'''
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Department": ["HR","IT","IT","HR","Sales","Sales"],
    "Salary": [30,50,60,35,45,55]
})

sns.set_theme(style="whitegrid")

sns.catplot(
    data=df,
    x="Department",
    y="Salary",
    kind="violin"
)

plt.show()
```

```python
#Join plot

'''
scatter + histogram(same time)

'''

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Department": ["HR","IT","IT","HR","Sales","Sales"],
    "Salary": [30,50,60,35,45,55]
})

sns.set_theme(style="whitegrid")

sns.jointplot(
    data=df,
    x="Department",
    y="Salary",
    kind="scatter"
)

plt.show()
```
