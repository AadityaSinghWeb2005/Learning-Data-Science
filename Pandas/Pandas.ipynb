```python
#Group By 
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Name': ['Aman', 'Riya', 'Rohit', 'Neha', 'Aman', 'Karan', 'Riya'],
    'City': ['Delhi', 'Mumbai', 'Delhi', 'Pune', 'Delhi', 'Mumbai', 'Mumbai'],
    'Marks': [85, 92, 78, 88, 85, 90, np.nan],
    'Age': [20, 21, 19, 22, 20, np.nan, 21]
})
# print(df[:])
df.groupby('City')['Age'].mean()

#Group + Multiple aggregration 

df.groupby('City')['Marks'].agg(['mean','max','min'])

#Groupby + count

df.groupby('City').size()

#Groupby Multiple columns

df.groupby(['City','Name'])['Marks'].mean()

#reset_index()
df.groupby('City')['Marks'].mean().reset_index()
'''
Jab hum normal filter karte hai tab index ulte sidhe hote obviously 
but reset_index() unhe wapis order mein kardeta hai 
'''

#Exercise 

df.groupby('City')['Marks'].mean()
df.groupby("City")['Marks'].max()

df.groupby('City').size()#.size() NaN values ko bhi count karta hai 
df.groupby('City')['Marks'].agg(['mean','max'])

df.groupby("City")["Marks"].mean().reset_index()

# mcq =[c,c,b,c,b]

```

```python
#Merge Join and concat 

import pandas as pd

students = pd.DataFrame({
    'StudentID': [101, 102, 103, 104],
    'Name': ['Aman', 'Riya', 'Rohit', 'Neha'],
    'City': ['Delhi', 'Mumbai', 'Delhi', 'Pune']
})

marks = pd.DataFrame({
    'StudentID': [101, 102, 104, 105],
    'Marks': [85, 92, 88, 90]
})
print(students)
print(marks)

# 1. Merge (Inner Join , Left JOin , RIght Join , Outer JOin )

#Inner JOin(default) 
pd.merge(students,marks,on='StudentID')
#sirf common student id aaenga 

#LEft Join 

pd.merge(students,marks,on='StudentID',how='left')
#left table ke sabhi rows likhenge aur agar value missing hui toh NaN ka use karenge 

#Right Join 

pd.merge(students,marks,on='StudentID',how='right')

#Outer Join 
pd.merge(students,marks,on='StudentID',how='outer')

#2 concat()

pd.concat([students,marks]) #Row wise concat

pd.concat([students,marks],axis=1)

#index based merge 
students.set_index('StudentID').join(marks.set_index('StudentID'))#jab index same ho join clearn hota hai 

#merge() is used to combine DataFrames using common columns,
# while concat() stacks DataFrames vertically or horizontally.


```
