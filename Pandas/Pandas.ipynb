```python
#Missing Values and Duplicate Handling

import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Name': ['Aman', 'Riya', 'Riya', 'Rohit', 'Neha'],
    'Marks': [85, np.nan, np.nan, 78, 88],
    'City': ['Delhi', 'Mumbai', 'Mumbai', np.nan, 'Pune']
})


df.isnull()

df.isnull().sum()



# df.dropna()
df.dropna(axis=1)#Very Dangerous

# df['Marks'] = df['Marks'].fillna(df['Marks'].mean())


df.ffill()
df.bfill()

df.duplicated()
df.drop_duplicates()

# df.drop_duplicates(subset=['Name'])
```
